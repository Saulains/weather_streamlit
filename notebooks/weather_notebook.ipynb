### Код для генерации файла

```python

# Реальные средние температуры (примерные данные) для городов по сезонам
seasonal_temperatures = {
    "New York": {"winter": 0, "spring": 10, "summer": 25, "autumn": 15},
    "London": {"winter": 5, "spring": 11, "summer": 18, "autumn": 12},
    "Paris": {"winter": 4, "spring": 12, "summer": 20, "autumn": 13},
    "Tokyo": {"winter": 6, "spring": 15, "summer": 27, "autumn": 18},
    "Moscow": {"winter": -10, "spring": 5, "summer": 18, "autumn": 8},
    "Sydney": {"winter": 12, "spring": 18, "summer": 25, "autumn": 20},
    "Berlin": {"winter": 0, "spring": 10, "summer": 20, "autumn": 11},
    "Beijing": {"winter": -2, "spring": 13, "summer": 27, "autumn": 16},
    "Rio de Janeiro": {"winter": 20, "spring": 25, "summer": 30, "autumn": 25},
    "Dubai": {"winter": 20, "spring": 30, "summer": 40, "autumn": 30},
    "Los Angeles": {"winter": 15, "spring": 18, "summer": 25, "autumn": 20},
    "Singapore": {"winter": 27, "spring": 28, "summer": 28, "autumn": 27},
    "Mumbai": {"winter": 25, "spring": 30, "summer": 35, "autumn": 30},
    "Cairo": {"winter": 15, "spring": 25, "summer": 35, "autumn": 25},
    "Mexico City": {"winter": 12, "spring": 18, "summer": 20, "autumn": 15},
}

# Сопоставление месяцев с сезонами
month_to_season = {12: "winter", 1: "winter", 2: "winter",
                   3: "spring", 4: "spring", 5: "spring",
                   6: "summer", 7: "summer", 8: "summer",
                   9: "autumn", 10: "autumn", 11: "autumn"}

# Генерация данных о температуре
def generate_realistic_temperature_data(cities, num_years=10):
    dates = pd.date_range(start="2010-01-01", periods=365 * num_years, freq="D")
    data = []

    for city in cities:
        for date in dates:
            season = month_to_season[date.month]
            mean_temp = seasonal_temperatures[city][season]
            # Добавляем случайное отклонение
            temperature = np.random.normal(loc=mean_temp, scale=5)
            data.append({"city": city, "timestamp": date, "temperature": temperature})

    df = pd.DataFrame(data)
    df['season'] = df['timestamp'].dt.month.map(lambda x: month_to_season[x])
    return df

# Генерация данных
data = generate_realistic_temperature_data(list(seasonal_temperatures.keys()))
data.to_csv('temperature_data.csv', index=False)
```

## 1. Вычислить скользящее среднее температуры с окном в 30 дней для сглаживания краткосрочных колебаний.

```python
# DataFrame.rolling(window, min_periods=None, center=False, win_type=None, on=None, axis=<no_default>, closed=None, step=None, method='single')
df["rolling_mean"] = (
    df.groupby("city")["temperature"]
    .rolling(window=30, min_periods=1)
    .mean()
    .reset_index(level=0, drop=True)
) # группировка, чтобы не захватить разные города
df.head()
```

В условии нет уточнения, поэтому рассчитаем среднюю температуру за каждый год отдельно и за последнее 10 лет для каждого сезона

```python
df["year"] = df["timestamp"].dt.year
```

```python
seasonal_yearly = (
    df.groupby(["city", "year", "season"])["temperature"]
      .mean()
      .reset_index()
)
seasonal_yearly.head()
```

```python
# Средняя температура по сезону (дублируется для каждой строки)
df["season_mean"] = df.groupby(["city", "season"])["temperature"].transform("mean")

# Проверим первые строки
df.head()
```

Стандартное отклонение:

```python
# Стандартное отклонение по сезону
df["season_std"] = df.groupby(["city", "season"])["temperature"].transform("std")

season_std = (
    df.groupby(["city", "season"])["temperature"]
      .agg(
          mean_temp="mean",
          std_temp="std",
          count="count"
      )
      .reset_index()
)
season_std.head()
```

## 3. Выявить аномалии, где температура выходит за пределы среднее±2𝜎.

```python
df["is_anomaly"] = (
    (df["temperature"] < df["rolling_mean"] - 2*df["season_std"]) |
    (df["temperature"] > df["rolling_mean"] + 2*df["season_std"])
)
df.head()
```

## 6. Сравнение скорости выполнения анализа с распараллеливанием и без него.

```python
import time
from joblib import Parallel, delayed
import pandas as pd
import matplotlib.pyplot as plt

start = time.time()

df["rolling_mean"] = (
    df.groupby("city")["temperature"]
      .rolling(window=30, min_periods=1)
      .mean()
      .reset_index(level=0, drop=True)
)

end = time.time()
time_sequential = end - start
print(f"Время без параллелизма: {time_sequential:.2f} секунд")

def compute_rolling(city_df):
    city_df = city_df.sort_values("timestamp").copy()
    city_df["rolling_mean"] = city_df["temperature"].rolling(window=30, min_periods=1).mean()
    return city_df

start = time.time()

# Получаем список DataFrame по городам
dfs = [group for _, group in df.groupby("city")]

# Распараллеливаем расчет
results = Parallel(n_jobs=-1)(delayed(compute_rolling)(g) for g in dfs)

# Собираем обратно в один DataFrame
df_parallel = pd.concat(results, ignore_index=True)

end = time.time()
time_parallel = end - start
print(f"Время с параллелизмом: {time_parallel:.2f} секунд")

methods = ["Sequential", "Parallel"]
times = [time_sequential, time_parallel]

plt.bar(methods, times, color=["blue", "green"])
plt.ylabel("Время, сек")
plt.title("Сравнение скорости расчёта rolling_mean")
plt.show()
```

В рамках данного датасета параллельный расчет не дает выигрыша по времени. Расходы на создание процессов и передачу данных превышают выигрыш от распараллеливания анализа.

## 3. Определить, является ли текущая температура нормальной, исходя из исторических данных для текущего сезона

Текущий сезон

```python
from datetime import datetime

def get_current_season():
    month = datetime.now().month

    if month in (12, 1, 2):
        return "winter"
    elif month in (3, 4, 5):
        return "spring"
    elif month in (6, 7, 8):
        return "summer"
    else:
        return "autumn"
current_season = get_current_season()
current_season
```

Исторические данные, проверка на аномалию

```python
def check_temperature_anomaly(df, city, current_temp):
    current_season = get_current_season()

    stats = df[
        (df["city"] == city) &
        (df["season"] == current_season)
    ]

    season_std = stats["season_std"].iloc[0]

    rolling_mean_current = (
        df[df["city"] == city]
          .sort_values("timestamp")
          ["rolling_mean"]
          .iloc[-1]
    )
    lower = rolling_mean_current - 2 * season_std
    upper = rolling_mean_current + 2 * season_std
    if lower <= current_temp <= upper:
        return "NORMAL"
    else:
        return "ANOMALY"

result = check_temperature_anomaly(df, CITY, temp)
print(result)
```

## 5. Попробуйте для получения текущей температуры использовать синхронные и асинхронные методы

Синхронные методы

```python
import requests

def get_temperature_sync(city, api_key):
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "q": city,
        "appid": api_key,
        "units": "metric"
    }

    resp = requests.get(url, params=params, timeout=10)
    data = resp.json()

    if resp.status_code == 200:
        return data["main"]["temp"]
    else:
        return None
```

```python
temp_sync = get_temperature_sync("Moscow", API_KEY)
print("Sync temperature:", temp_sync)
```

Асинхронные методы

```python
import aiohttp
import asyncio

async def get_temperature_async(city, api_key):
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "q": city,
        "appid": api_key,
        "units": "metric"
    }

    async with aiohttp.ClientSession() as session:
        async with session.get(url, params=params) as resp:
            data = await resp.json()
            if resp.status == 200:
                return data["main"]["temp"]
            else:
                return None
```

```python
temp_async = await get_temperature_async("Moscow", API_KEY)
print("Async temperature:", temp_async)
```

```python
import time

# sync
start = time.time()
temp_sync = get_temperature_sync("Moscow", API_KEY)
sync_time = time.time() - start

# async
start = time.time()
temp_async = await get_temperature_async("Moscow", API_KEY)
async_time = time.time() - start

print("Sync temp:", temp_sync, "time:", sync_time)
print("Async temp:", temp_async, "time:", async_time)
```

Асинхронный подход, конечно, эффективнее, но все же его целесообразно использовать при одновременном запросе данных для большого числа городов, однако в рамках данной задачи предпочтительным является синхронный метод.
